```python
import gzip
import pickle
import numpy as np
from neural_network import MLPClassifier
```

```python
with gzip.open("../data/mnist.pkl.gz") as file:
    train_set, valid_set, test_set = pickle.load(file, encoding="latin1")

X, y = train_set
n, d = X.shape
print(f"n = {n}, d = {d}")
```

```python
model = MLPClassifier(
    hidden_layer_sizes=[256, 128],
    learning_rate=0.01,
    verbose=True,
)
model.fit(X, y)
```

```python
X_test, y_test = test_set

# Compute training error
yhat = model.predict(X)
train_error = np.mean(yhat != y)
print(f"Training error = {train_error}")

# Compute test error
yhat = model.predict(X_test)
test_error = np.mean(yhat != y_test)
print(f"Test error = {test_error}")
```

```python
with open("../models/digit_classifier.pkl",'wb') as f:
    pickle.dump(model, f)
```
